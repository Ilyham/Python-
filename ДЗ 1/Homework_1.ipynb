```python
purchases = {}
i = 0
with open('purchase_log.txt', encoding='utf-8') as f:
    for line in f:
        
        stroka = json.loads(line)
        #print(stroka, type(stroka))
        
        user_id = stroka['user_id']
        #print(user_id, type(user_id))
        
        category = stroka ['category']
        #print(category, type(category))

        purchases[user_id] = category
        
first_five = list(purchases.items())[:5]
for user_id, category in first_five:
    print(user_id, category)
```
